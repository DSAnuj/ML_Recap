```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")
```

```python
# Now let's read the df from the csv folder
df = pd.read_csv("../csv/home_price_feature_selection.csv")
df.head()
```

```python
df.isna().sum()
```

```python
# Now let's create train and test data
```

```python
x = df.drop('sale_price', axis=1)
y = df['sale_price']
x.shape, y.shape
```

### We will first test by creating backward_feature_elimination process

```python
from mlxtend.feature_selection import SequentialFeatureSelector as sfs
from sklearn.linear_model import LinearRegression
```

```python
lreg = LinearRegression()
sfs1 = sfs(lreg, k_features=6, forward=False, verbose=1, scoring="neg_mean_squared_log_error")
```

```python
sfs1 = sfs1.fit(x,y)
```

```python
feat_names = list(sfs1.k_feature_names_)
```

```python
feat_names
```

```python
new_data = df[feat_names]
```

```python
new_data['sale_price'] = df['sale_price']
```
