```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")
```

```python
# read the dataframe
df = pd.read_csv("../csv/home_price_feature_selection.csv")
df.head()
```

```python
df.isna().sum()
```

```python
x = df.drop("sale_price", axis=1)
y = df['sale_price']
x.shape, y.shape
```

```python
from mlxtend.feature_selection import SequentialFeatureSelector as sfs
from sklearn.linear_model import LinearRegression
```

```python
lreg = LinearRegression()
sfs1 = sfs(lreg, k_features=6, forward=True, verbose=2, scoring='neg_mean_squared_error')
```

```python
sfs1 = sfs1.fit(x,y)
```

```python
feat_names = list(sfs1.k_feature_names_)
print(feat_names)
```

```python
new_data = df[feat_names]
new_data['sale_price'] = df['sale_price']
new_data.head()
```

```python
new_data.shape, df.shape
```
